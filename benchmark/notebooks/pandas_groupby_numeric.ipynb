# Group means on a mixed-type table

```python
import numpy as np
import pandas as pd
```

```python
sales = pd.DataFrame({'region': ['n', 'n', 's', 's', 's'], 'rep': ['ali', 'bo', 'cy', 'di', 'ed'], 'units': [10, 14, 7, 9, 11], 'price': [2.5, 2.0, 3.0, 3.5, 2.5]})
```

```python
print(sales.groupby('region').mean().round(3).to_string())
```
