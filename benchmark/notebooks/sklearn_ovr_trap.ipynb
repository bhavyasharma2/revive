# One-vs-rest classifier (dropping the old flag changes results)

```python
from sklearn.datasets import make_classification
from sklearn.linear_model import LogisticRegression
from sklearn.multiclass import OneVsRestClassifier
```

```python
X, y = make_classification(n_samples=300, n_features=8, n_informative=5, n_classes=4, n_clusters_per_class=1, random_state=11)
```

```python
clf = LogisticRegression(multi_class='ovr', max_iter=2000).fit(X[:200], y[:200])
print(round(clf.score(X[200:], y[200:]), 4))
print(clf.predict_proba(X[200:203]).round(3).tolist())
```
