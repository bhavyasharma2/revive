# Label lookups with .ix and DataFrame.lookup

```python
import numpy as np
import pandas as pd
```

```python
scores = pd.DataFrame({'math': [91, 78, 85], 'physics': [88, 95, 70], 'chem': [72, 81, 90]}, index=['ana', 'ben', 'chloe'])
```

```python
print(scores.ix['ben', 'physics'])
print(scores.ix[0].tolist())
```

```python
favourite = ['physics', 'math', 'chem']
print(scores.lookup(scores.index, favourite).tolist())
```
