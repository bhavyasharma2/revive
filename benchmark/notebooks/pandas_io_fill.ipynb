# Reading whitespace data and forward-filling gaps

```python
import numpy as np
import pandas as pd
import io
```

```python
raw = 'day temp\n1 21.5\n2 NaN\n3 23.0\n4 NaN\n5 24.5'
df = pd.read_csv(io.StringIO(raw), delim_whitespace=True)
print(df.shape)
```

```python
print(df['temp'].fillna(method='ffill').tolist())
```
