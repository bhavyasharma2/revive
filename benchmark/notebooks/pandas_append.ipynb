# Collecting experiment results row by row

```python
import numpy as np
import pandas as pd
```

```python
results = pd.DataFrame(columns=['run', 'lr', 'loss'])
for i, lr in enumerate([0.1, 0.01, 0.001]):
    row = {'run': i, 'lr': lr, 'loss': round(1 / (1 + 50 * lr), 4)}
    results = results.append(row, ignore_index=True)
print(results.to_string())
```

```python
tags = pd.Series(['baseline', 'tuned'])
tags = tags.append(pd.Series(['final']), ignore_index=True)
print(tags.tolist())
```
