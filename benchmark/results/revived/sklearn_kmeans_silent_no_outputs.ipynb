# Customer segments with KMeans (runs fine, wrong numbers)

```python
from sklearn.cluster import KMeans
from sklearn.datasets import make_blobs
```

```python
X, _ = make_blobs(n_samples=300, centers=6, cluster_std=2.2, random_state=4)
```

```python
km = KMeans(n_clusters=6, random_state=0, n_init=10).fit(X)
print(round(km.inertia_, 2))
```
