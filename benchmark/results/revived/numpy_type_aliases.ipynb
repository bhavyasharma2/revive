# Feature scaling with NumPy type aliases

```python
import numpy as np
import pandas as pd
```

```python
rng = np.random.RandomState(1)
ages = rng.randint(18, 80, 50)
income = rng.normal(52000, 9000, 50)
```

```python
scaled = (income - income.mean()).astype(float) / income.std()
print(round(float(scaled[:5].sum()), 5))
```

```python
buckets = (ages // 10).astype(int)
print(np.bincount(buckets).tolist())
```

```python
is_senior = (ages > 60).astype(bool)
print(int(is_senior.sum()))
```
