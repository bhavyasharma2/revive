# Regression error report

```python
import numpy as np
from sklearn.linear_model import Ridge
from sklearn.datasets import make_regression
```

```python
X, y = make_regression(n_samples=120, n_features=4, noise=12.0, random_state=3)
model = Ridge(alpha=1.0).fit(X[:90], y[:90])
pred = model.predict(X[90:])
```

```python
from sklearn.metrics import mean_squared_error
print(round(np.sqrt(mean_squared_error(y[90:], pred)), 4))
```
