# Robust statistics with mad and applymap

```python
import numpy as np
import pandas as pd
```

```python
df = pd.DataFrame({'a': [1.0, 2.5, 3.0, 10.0], 'b': [4.0, 4.5, 5.5, 6.0]})
```

```python
print((df - df.mean()).abs().mean().round(4).to_dict())
```

```python
print(df.map(lambda v: f'{v:.1f}').iloc[0].tolist())
```
