# Weekly features from timestamps

```python
import numpy as np
import pandas as pd
```

```python
dates = pd.Series(pd.date_range('2021-03-01', periods=6, freq='5D'))
```

```python
print(dates.dt.isocalendar().week.tolist())
```

```python
ts = pd.Series(range(6), index=dates.values)
print(ts.loc[:ts.index[0] + pd.Timedelta('11D')].tolist())
```
