# Array utilities renamed in NumPy 2

```python
import numpy as np
import pandas as pd
```

```python
growth = np.array([1.02, 1.05, 0.98, 1.10])
print(round(float(np.prod(growth)), 6))
print(np.round(np.cumprod(growth), 4).tolist())
```

```python
print(bool(np.all(growth > 0.9)))
print(np.isin([1, 5, 9], [5, 9, 12]).tolist())
```

```python
x = np.linspace(0, 1, 11)
print(round(float(np.trapezoid(x ** 2, x)), 6))
print(np.vstack([x[:2], x[-2:]]).shape)
```
